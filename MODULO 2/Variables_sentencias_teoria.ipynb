## 1. Variables y Diagramas de Estado
* **Variable:** Es un nombre que hace referencia a un valor almacenado en memoria.
* **Sentencia de asignación:** Crea una variable y le asigna un valor usando el operador `=`. No genera salida visible al ejecutarse.
* **Diagrama de estado:** Representación visual donde el nombre de la variable apunta mediante una flecha hacia su valor actual.

````python
n = 17                  # Entero (int)
pi = 3.141592653589793  # Punto flotante (float)
message = 'Hola mundo'  # Cadena de texto (str)

message
````

## 2. Reglas para Nombrar Variables
* **Permitido:** Letras (preferiblemente minúsculas), números y guiones bajos (`_`), este último usado para separar palabras (*snake_case*, ej. `velocidad_luz`).
* **Prohibido (genera `SyntaxError`):**
  * Empezar con un número (`76trombones`).
  * Usar signos de puntuación o caracteres especiales (`million!`, `precio$`).
  * Usar **palabras clave** (*keywords*) reservadas por Python:

```text
False      await      else       import     pass
None       break      except     in         raise
True       class      finally    is         return
and        continue   for        lambda     try
as         def        from       nonlocal   while
assert     del        global     not        with
async      elif       if         or         yield
```


## 3. Módulos y la Sentencia `import`
* **Módulo:** Archivo que agrupa variables y funciones relacionadas.
* **Importación:** Se cargan con la sentencia `import`.
* **Operador punto (`.`):** Permite acceder a los elementos internos de un módulo (`nombre_modulo.elemento`).

````python
import math

print(math.pi)        # Variable pi del módulo math
print(math.sqrt(25))  # Función raíz cuadrada -> 5.0
print(math.pow(5, 2))  # Función potencia -> 25.0 (equivale a 5 ** 2)
````

## 4. Expresiones vs. Sentencias
| Concepto | Definición | Acción | Ejemplo |
| :--- | :--- | :--- | :--- |
| **Expresión** | Combinación de valores, variables, operadores y llamadas a funciones que produce un resultado. | Se **evalúa** | `19 + n + round(math.pi)` |
| **Sentencia** | Unidad de código que ejecuta una orden o tiene un efecto, pero no devuelve un valor por sí misma. | Se **ejecuta** | `n = 17` o `import math` |

## 5. La Función `print` y los Argumentos
* **`print()`:** Muestra valores en pantalla. Si recibe varios argumentos separados por comas, los imprime separados por un espacio.
* **Argumento:** Es el valor o expresión que se pasa entre paréntesis al llamar a una función.
  * **Obligatorios:** `math.pow(5, 2)` requiere exactamente 2 argumentos.
  * **Opcionales:** `round(math.pi, 3)` redondea a 3 decimales; `int('101', 2)` convierte texto en base 2 a entero (`5`) ya que lo interpreta como binario.
  * **Errores comunes (`TypeError`):** Pasar más argumentos de los permitidos, menos de los necesarios o un tipo de dato incompatible (ej. `math.sqrt('123')`).

````python
print('El valor de pi es:', math.pi)
print('Binario 101 a decimal:', int('101', 2))
print('Binario 110 a decimal:', int('110', 2))
print('Pi redondeado a 3 decimales:', round(math.pi, 3))
````

## 6. Comentarios
* Se escriben usando el símbolo `#`. Python ignora todo lo que esté a su derecha en esa misma línea.
* **Regla de oro:** Úsalos para explicar **por qué** hace algo el código o aclarar unidades/contexto, no para repetir lo que ya es obvio.

## 7. Debugging: Los 3 Tipos de Errores
1. **Error de sintaxis (`SyntaxError`):** Fallo en la estructura o reglas del lenguaje. Impide que el programa arranque (ej. `1var = 5`).
2. **Error en tiempo de ejecución / Excepción:** El código tiene buena sintaxis, pero falla al ejecutarse por una operación imposible (ej. `'126' / 3` genera `TypeError`, o importar un módulo inexistente genera `ModuleNotFoundError`).
3. **Error semántico:** El programa se ejecuta sin fallos ni advertencias, pero el resultado es incorrecto por un fallo de lógica (ej. calcular el promedio con `1 + 3 / 2` en lugar de `(1 + 3) / 2`).
