````python
# Cálculo de número de euler
import math
calculo_euler_uno = math.e **3
calculo_euler_dos = math.pow(math.e, 3)
calculo_euler_tres = math.exp(3)
print('El cálculo de euler usando e**3 es:', round(calculo_euler_uno, 90), '. Usando pow(e, 3) es:', round(calculo_euler_dos, 90), '. Usando exp(3) es:', round(calculo_euler_tres, 90)) # Si veo diferencias en los últimos dígitos
````

### Ejercicio 5: Reto Integrador (Física y Comentarios)
Un proyectil se lanza verticalmente hacia arriba con una velocidad inicial $v_0 = 25\text{ m/s}$ desde una altura inicial $y_0 = 2\text{ m}$. La fórmula de su altura en un tiempo $t$ es:
$$y = y_0 + v_0 \cdot t - \frac{1}{2} g \cdot t^2$$

Escribe un script completo en Python que:
1. Defina variables con nombres descriptivos en *snake_case* para la altura inicial, velocidad inicial, gravedad ($9.81\text{ m/s}^2$) y el tiempo ($t = 3.5\text{ segundos}$).
2. Incluya comentarios explicando las unidades de cada variable.
3. Calcule la altura final y la redondee a 2 decimales.
4. Muestre el resultado final acompañado de un mensaje claro usando `print()`.

````python
# Calcular la altura final del proyectil
velocidad_inicial = 25 #m/s
altura_inicial = 2 #m
gravedad = 9.81 #m/s²
tiempo= 3.5 #s
altura_final = altura_inicial + velocidad_inicial*tiempo - 0.5 * gravedad * tiempo **2
print('La altura final del proyectil es:', round(altura_final, 2), 'm') # 29.41 m
````
