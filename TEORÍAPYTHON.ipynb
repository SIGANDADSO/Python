# String

Python puede representar secuencias de letras, que se llaman strings porque las letras están ensartadas como cuentas en un collar. Para escribir un string, podemos poner una secuencia de letras entre comillas rectas. En este caso, puedes usar comillas simples('') y comillas dobles (" ").


```python
#1. Ejemplo: Comillas simples.
print('Hola mundo')
```

```python
#2. Ejemplo: Comillas dobles
print("Hola mundo")
```

```python
#3. Ejemplo: Las comillas dobles facilitan escribir un string que contiene un apóstrofo que una comilla simple.
print("It's very small")
```

```python
#4. Ejemplo: Para realizar una concatenación es con el operador + para unir texto .
print("It's very small" + '. Bye!')
```

```python
#5. Ejemplo: Hacer varias copias es con el operador *.
print(('Bye!')*4)
```

```python
#6. Ejemplo: calcular la longitud del string.
len("small")
```
