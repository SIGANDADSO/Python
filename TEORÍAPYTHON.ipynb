# String

Python puede representar secuencias de letras, que se llaman strings porque las letras están ensartadas como cuentas en un collar. Para escribir un string, podemos poner una secuencia de letras entre comillas rectas. En este caso, puedes usar comillas simples('') y comillas dobles (" ").


```python
#1. Ejemplo: Comillas simples.
print('Hola mundo')
```

```python
#2. Ejemplo: Comillas dobles
print("Hola mundo")
```

```python
#3. Ejemplo: Las comillas dobles facilitan escribir un string que contiene un apóstrofo que una comilla simple.
print("It's very small")
```

```python
#4. Ejemplo: Para realizar una concatenación es con el operador + para unir texto .
print("It's very small" + '. Bye!')
```

```python
#5. Ejemplo: Hacer varias copias es con el operador *.
print(('Bye!')*4)
```

```python
#6. Ejemplo: calcular la longitud del string.
len("small")
```

# Valores y tipos
Hasta ahora hemos visto valores enteros, decimales y string en el desarrollo de ejemplos. En esta sección usaremos la función type() para determinar el tipo de cualquier valor que se le asigne.

```python
#1. Ejemplo: determinar el tipo de valor a 34.9
type(34.9)
```

```python
#2. Ejemplo: determinar el tipo de valor a 34
type(34)
```

```python
#3. Ejemplo: determinar el tipo de valor "carlos"
type("carlos")
```

Hemos visto que hay tipos float, int y string (str)  para los datos evaluados. Ahora bien, algunas veces tenemos números entre comillas sinples o dobles que los hace tipo string y toca convertilos en float o int.

```python
#4. Ejemplo: convertir en número entero el dato "348"
int("348")
```

```python
#5. Ejemplo: convertir en número decimal el dato "34.8"
float("34.8")
```
