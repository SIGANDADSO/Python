```python
#¿Cuántos segundos hay en 42 minutos y 42 segundos?

a = 42*60 # cambiar a segundos
totalSegundos = a + 42 # sumar segundos
print(totalSegundos, "segundos")
```

```python
#¿Cuántas millas hay en 10 kilómetros? Pista: hay 1,61 kilómetros en una milla.

millas = 10/1.61
print(round(millas), "mi")
```

```python
#Si corres una carrera de 10 kilómetros en 42 minutos y 42 segundos, ¿cuál es tu ritmo medio en segundos por milla?
segundosTotales = 42*60 + 42 # segundos
ritmoMedio = segundosTotales/millas
print(ritmoMedio, "s/mi")
```

```python
#¿Cuál es tu ritmo medio en minutos y segundos por milla?
distancia = 20 #km
tiempoSegundo = 45*60 # segundos
distanciaMillas = distancia/1.61
ritmoCarreraMedio = tiempoSegundo/distanciaMillas
print("Ritmo = ", round(ritmoCarreraMedio), "s/mi")
print(round(ritmoCarreraMedio), "s/mi =", round(ritmoCarreraMedio/60), "minutos y", round(ritmoCarreraMedio%60), "segundos por milla")
```

```python
#¿Cuál es tu velocidad media en millas por hora?
d1 = 10/1.61 #mi
d2 = 20/1.61 #mi
t = 2 # horas
d3 = d2-d1
velocidad = d3/t
print(round(velocidad), "mi/h")
```
